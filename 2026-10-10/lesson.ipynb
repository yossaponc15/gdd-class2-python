## 1. รับข้อมูลตาราง
พิมพ์ `2 3` แล้วตามด้วย 2 บรรทัด เช่น `1 2 3` และ `4 5 6`

```python
n, m = map(int, input().split())
grid = [list(map(int, input().split())) for _ in range(n)]
print(grid)
print("ผลรวมแถว:", [sum(row) for row in grid])
print("ผลรวมคอลัมน์:", [sum(grid[r][c] for r in range(n)) for c in range(m)])
```

## 2. แยก function + assert

```python
def solve(arr):
    return max(arr) - min(arr)

assert solve([3, 9, 1]) == 8
assert solve([5]) == 0
assert solve([-2, -7]) == 5
print("ผ่านทุก test ✅")
```

## 3. ความเร็ว ⏱️
เทียบเวลา: วน 2 ชั้น vs prefix sum (ลองเพิ่ม N ดู)

```python
import time, random
N, Q = 2000, 2000
arr = [random.randint(1, 100) for _ in range(N)]
queries = [(1, N)] * Q

t = time.time()
for l, r in queries:
    s = 0
    for i in range(l - 1, r):
        s += arr[i]
print("วนบวกทุกครั้ง:", round(time.time() - t, 3), "วินาที")

t = time.time()
pre = [0]
for x in arr:
    pre.append(pre[-1] + x)
for l, r in queries:
    s = pre[r] - pre[l - 1]
print("prefix sum:", round(time.time() - t, 3), "วินาที")
```

## 4. Sliding Window

```python
arr = [2, 1, 5, 1, 3, 2]
k = 3
window = sum(arr[:k])
best = window
for i in range(k, len(arr)):
    window += arr[i] - arr[i - k]
    print(f"หน้าต่าง {arr[i-k+1:i+1]} = {window}")
    best = max(best, window)
print("ดีที่สุด:", best)
```

## 5. ตาราง + เพื่อนบ้าน

```python
grid = ["#..",
        ".#.",
        "..."]
n, m = 3, 3
dirs = [(-1, 0), (1, 0), (0, -1), (0, 1)]
r, c = 1, 2
for dr, dc in dirs:
    nr, nc = r + dr, c + dc
    if 0 <= nr < n and 0 <= nc < m:
        print((nr, nc), grid[nr][nc])
    else:
        print((nr, nc), "นอกตาราง")
```
