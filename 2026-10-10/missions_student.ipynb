## ข้อ 8: 🏆 ระบบคลังกาแล็กซี

ระบบคลังรับคำสั่ง N คำสั่ง:
- `ADD ดาว x` — รับพัสดุไปดาวนั้นเข้าคลัง x ชิ้น
- `SEND ดาว x` — ส่งพัสดุออก x ชิ้น ถ้าในคลังมีไม่พอให้แสดง `FAIL` และไม่ส่ง ไม่งั้นแสดง `OK`
- `QUERY ดาว` — แสดงจำนวนพัสดุของดาวนั้นในคลัง (ไม่มี = 0)
- `REPORT` — แสดงทุกดาวที่ยังมีพัสดุ > 0 เรียงตามจำนวน **มาก→น้อย** ถ้าเท่ากันเรียงชื่อ A→Z รูปแบบ `ดาว จำนวน` บรรทัดละดาว (ถ้าไม่มีเลย แสดง `EMPTY`)

**ข้อมูลนำเข้า** บรรทัดแรก N (`1 ≤ N ≤ 100,000`) ตามด้วยคำสั่ง N บรรทัด

**ตัวอย่าง Input**
```text
9
ADD mars 5
ADD venus 3
ADD mars 2
SEND mars 4
SEND pluto 1
ADD pluto 10
SEND venus 5
QUERY mars
REPORT
```
**ตัวอย่าง Output**
```text
OK
FAIL
FAIL
3
pluto 10
mars 3
venus 3
```

> 💡 แยก function ต่อคำสั่ง: `add`, `send`, `query`, `report`
