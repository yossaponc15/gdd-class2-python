# 🅱️ ช่วง B — คัดกรอง (filter) → Level 16

**ตัวสะสมที่เป็น list:** เริ่มจาก `[]` แล้ว `append` เฉพาะตัวที่ผ่านเงื่อนไข

````python
def strong_satellites(satellites, minimum):
    result = []                         # ① list ว่าง
    for s in satellites:
        if s["signal"] >= minimum:      # ② เงื่อนไข
            result.append(s)            # ③ เก็บทั้ง dict
    return result                       # ④ นอก loop

for s in strong_satellites(satellites, 60):
    print(s["name"])
````

> 💡 ถ้าเทียบกับ **ข้อความ** ที่ผู้ใช้พิมพ์ ให้ `clean()` คำนั้น **ก่อน loop**
> ⚠️ บาง dict อาจ **ไม่มี key** ที่จะเช็ก → ใช้ `s.get("key")` จะได้ `None` แทนที่จะ `KeyError`

# 🅲 ช่วง C — เทียบกับผลของฟังก์ชันอื่น → Level 17–18

ฟังก์ชันตรวจความพร้อมส่วนใหญ่มีหน้าตาแบบนี้:
```python
def something_is_ready(data):
    return ของที่มี >= ฟังก์ชันที่คำนวณของที่ต้องใช้(data)
```
- ผลของ `>=` เป็น `True/False` อยู่แล้ว → **return ได้เลย**
- "เท่ากันถือว่าพอ" → ใช้ `>=` ไม่ใช่ `>`

````python
def power_needed(satellite):
    return satellite["signal"] // 10 + 2        # สูตรคำนวณอยู่ที่เดียว

def power_is_ready(satellite, battery):
    return battery >= power_needed(satellite)   # เรียกใช้ ไม่เขียนสูตรซ้ำ

sky = {"name": "Sky Eye", "signal": 70}
print(power_needed(sky))
print(power_is_ready(sky, 9))
print(power_is_ready(sky, 5))
````
