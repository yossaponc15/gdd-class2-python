# 🅰️ ช่วง A — Function + `return`

## A1. `print` ≠ `return`
- `print` = **แสดงบนจอ** แล้วจบ ค่าที่ได้คือ `None`
- `return` = **ส่งค่ากลับ** ให้โค้ดที่เรียกเอาไปใช้ต่อ

````python
def fuel_print(distance):
    print((distance + 9) // 10)

def fuel_return(distance):
    return (distance + 9) // 10

a = fuel_print(240)       # แสดง 24 บนจอ แต่ a ได้ None
b = fuel_return(240)      # ไม่แสดงอะไร แต่ b ได้ 24
print("a =", a)
print("b =", b, "→ เอาไปคำนวณต่อได้:", b * 2)
````

> 💡 โจทย์ใน notebook หลักเขียนว่า **"คืนค่า"** = ต้องใช้ `return` เสมอ

## A2. ฟังก์ชันเรียกฟังก์ชัน (อย่าเขียนสูตรซ้ำ!)

## A3. ฟังก์ชันที่คืน `True` / `False` ⭐ (ใช้ใน Level 10)

เทคนิค: **แตกเงื่อนไขเป็น boolean ย่อยทีละตัว** ตั้งชื่อให้อ่านรู้เรื่อง แล้วรวมด้วย `and`

````python
def can_join(crew):
    old_enough = crew["age"] >= 18
    good_role = crew["role"] not in ["Pirate", "Spy"]
    return old_enough and good_role          # คืน True/False ได้เลย ไม่ต้อง if

print(can_join(create_crew("Nova", "Engineer", 19)))
print(can_join(create_crew("Rex", "Pirate", 30)))
print(can_join(create_crew("Kit", "Pilot", 15)))
````

> ⚠️ ไม่ต้องเขียน `if x: return True else: return False` — `return x` ได้เลย

# 🅱️ ช่วง B — List ของ Dict + ตัวสะสม

## B1. list ที่ข้างในเป็น dict
แต่ละรอบของ `for` ตัวแปรจะได้ **dict หนึ่งตัว** → ใช้ `[...]` ดึงค่า

````python
samples = [
    {"name": "Ice Rock", "mass": 4, "rare": False},
    {"name": "Star Dust", "mass": 2, "rare": True},
    {"name": "Moon Gem", "mass": 7, "rare": True},
]

for s in samples:
    print(s["name"], "หนัก", s["mass"])
````

## B2. รูปแบบตัวสะสม (accumulator) 🧺

```python
total = 0              # 1) ตั้งค่าเริ่มต้น "นอก" loop
for item in items:     # 2) วนทีละชิ้น
    total += ...       # 3) บวกเพิ่ม
return total           # 4) return "หลัง" loop (ย่อหน้าเดียวกับ for)
```

````python
def total_mass(samples):
    total = 0
    for s in samples:
        total += s["mass"]
    return total

def count_rare(samples):
    count = 0
    for s in samples:
        if s["rare"]:              # นับเฉพาะที่ตรงเงื่อนไข
            count += 1
    return count

print(total_mass(samples))
print(count_rare(samples))
````

> ⚠️ จุดพลาดบ่อย: วาง `return` **ใน** loop → ฟังก์ชันจบตั้งแต่รอบแรก

## B3. เรียกฟังก์ชันกับทุกชิ้น + เพิ่มของลง list

````python
def sample_value(s):
    value = s["mass"] * 10
    if s["rare"]:
        value += 50
    return value

def total_value(samples):
    total = 0
    for s in samples:
        total += sample_value(s)      # ใช้ฟังก์ชันเดิม ไม่เขียนสูตรซ้ำ
    return total

print(total_value(samples))

samples.append({"name": "Dark Ore", "mass": 3, "rare": False})   # เพิ่มของเข้า list
print(len(samples), "ชิ้น")
````

# 🅲 ช่วง C — ค้นหา & แก้ไข

## C1. ฟังก์ชันค้นหา: เจอคืน dict · ไม่เจอคืน `None`

````python
crew_list = [
    create_crew("nova", "pilot", 19),
    create_crew("rex", "engineer", 25),
]

def find_crew(crew_list, name):
    target = clean(name)              # จัดคำค้นครั้งเดียว ก่อน loop
    for c in crew_list:
        if c["name"] == target:
            return c                  # เจอแล้ว → ออกจากฟังก์ชันทันที
    return None                       # วนครบแล้วไม่เจอ (อยู่ "นอก" loop!)

print(find_crew(crew_list, "  REX "))
print(find_crew(crew_list, "Zed"))
````

> ⚠️ ถ้าเขียน `else: return None` **ใน** loop → จะตรวจแค่ชิ้นแรกแล้วจบ

## C2. ใช้ผลค้นหาต่อ: ตรวจ `is None` ก่อนเสมอ
dict ที่ได้จากการค้นหา **คือตัวเดียวกับใน list** แก้ตรงนี้ ใน list ก็เปลี่ยนด้วย

````python
def promote(crew_list, name, new_role):
    found = find_crew(crew_list, name)    # ใช้ฟังก์ชันค้นหาเดิม
    if found is None:
        return False
    found["role"] = new_role               # แก้ค่าใน dict
    return True

print(promote(crew_list, "nova", "Captain"))
print(promote(crew_list, "zed", "Captain"))
print(crew_list[0])                        # ใน list เปลี่ยนตามด้วย!
````
