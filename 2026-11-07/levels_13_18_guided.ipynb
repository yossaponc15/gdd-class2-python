**เริ่มต้น:** รัน cell 🔧 ข้างล่างก่อน (เป็นฟังก์ชัน Level 1–12 ที่เราทำมาแล้ว)

### 🔧 ฟังก์ชัน Level 1–12 (รันก่อนเริ่ม)

````python
# 🔧 Level 1–12
def normalize_text(text):
    return text.strip().title()


def create_pilot(name, callsign):
    return {
        "name": normalize_text(name),
        "callsign": normalize_text(callsign),
    }


def create_ship(name, fuel, capacity):
    if fuel < 0:
        fuel = 0
    return {
        "name": normalize_text(name),
        "fuel": fuel,
        "capacity": capacity,
    }


def create_planet(name, distance, is_open):
    return {
        "name": normalize_text(name),
        "distance": distance,
        "is_open": is_open,
    }


def create_mission(pilot, ship, planet, budget):
    return {
        "pilot": pilot,
        "ship": ship,
        "planet": planet,
        "budget": budget,
        "packages": [],
        "launched": False,
    }


def calculate_required_fuel(mission):
    distance = mission["planet"]["distance"]
    return (distance + 9) // 10


def create_package(name, weight, package_type):
    return {
        "name": normalize_text(name),
        "weight": weight,
        "type": normalize_text(package_type),
        "status": "รอตรวจสอบ",
    }


def calculate_package_cost(package):
    cost = package["weight"] * 15
    if package["type"] == "Fragile":
        cost += 20
    return cost


def total_cargo_weight(mission):
    total = 0
    for package in mission["packages"]:
        total += package["weight"]
    return total


def is_package_allowed(mission, package):
    weight_ok = package["weight"] <= 20
    type_ok = package["type"] not in ["Weapon", "Explosive"]
    capacity_ok = (
        total_cargo_weight(mission) + package["weight"]
        <= mission["ship"]["capacity"]
    )
    return weight_ok and type_ok and capacity_ok


def add_package(mission, package):
    if is_package_allowed(mission, package):
        package["status"] = "พร้อมส่ง"
        mission["packages"].append(package)
        return True
    return False


def total_shipping_cost(mission):
    total = 0
    for package in mission["packages"]:
        total += calculate_package_cost(package)
    return total
````
